# Object detection

Find and locate objects in images or video with bounding boxes; count and track them.

Two ways to start, from Hello Model: https://sayhellomodel.com/models/object-detection/

**Your data:** the code uses example file names (such as `data.csv`). Upload your own file with the Files panel on the left, or change the name in the code. Run the cells from the top.

## Quick start

Pretrained YOLO / open-vocabulary detector (YOLO-World, Grounding DINO): Detect common objects or describe new ones in words — no training needed.

```python
%pip install -q ultralytics
```

```python
# pip install ultralytics
from ultralytics import YOLO
model = YOLO("yolo11n.pt")            # pretrained on 80 COCO classes
results = model("store_entrance.jpg")
people = [b for b in results[0].boxes if int(b.cls) == 0]
print("people detected:", len(people))
```

## Train your own model

Fine-tuned YOLO (Ultralytics) on your labeled images: Excellent speed/accuracy, simple training CLI, exports to ONNX/TensorRT/CoreML.

```python
%pip install -q ultralytics
```

```python
# pip install ultralytics
from ultralytics import YOLO

# data.yaml lists train/val image folders and class names
model = YOLO("yolo11s.pt")
model.train(data="data.yaml", epochs=100, imgsz=640, batch=16)
metrics = model.val()
print("mAP50:", metrics.box.map50)
model.export(format="onnx")   # or "engine" (TensorRT), "coreml", "tflite"
```
